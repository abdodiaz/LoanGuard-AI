1️⃣ Préparation des données

Charger le dataset (train.csv ou dataset.csv).

```python
import pandas as pd
```

charge dataset

```python
df = pd.read_csv('../data/data-entrainement.csv')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

Traiter les valeurs manquantes :
Catégoriques → valeur la plus fréquente

```python
cat_cols = ['Gender', 'Married', 'Dependents', 'Education', 'Self_Employed', 'Property_Area', 'Credit_History']
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
print(df[cat_cols].isnull().sum())
```

Numériques → valeur précédente ou médiane

```python
num_cols = ['ApplicantIncome', 'CoapplicantIncome', 'LoanAmount', 'Loan_Amount_Term']
```

```python
for col in num_cols:
    valeur_mediane = df[col].median()
    df[col] = df[col].fillna(valeur_mediane)
```

```python
print(df.isnull().sum())
```

Encoder les variables catégoriques avec LabelEncoder.

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
cols_to_encode = ['Gender', 'Married', 'Dependents', 'Education', 'Self_Employed', 'Property_Area', 'Loan_Status']
for col in cols_to_encode:
    df[col] = le.fit_transform(df[col])

df.head()
```

Séparer features et target.

```python

y = df['Loan_Status']

X = df.drop(['Loan_Status', 'Loan_ID'], axis=1)

print("Taille dial X (Features):", X.shape)
print("Taille dial y (Target):", y.shape)
```

Split train/test (80/20, stratifié).

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)


print("Train:", X_train.shape[0])
print("Test:", X_test.shape[0])
```

2️⃣ Modélisation et Évaluation

Entraîner Logistic Regression sur les features sélectionnées.

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(solver='liblinear', max_iter=2000)
model.fit(X_train, y_train)

```

Évaluer le modèle sur le set test avec au minimum :
Accuracy (précision globale)
Confusion Matrix
Précision, Rappel et F1-score

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# 1. L-modèle kiy-jawb 3la l-mtihaan
y_pred = model.predict(X_test)

# 2. Accuracy (Précision globale)
acc = accuracy_score(y_test, y_pred)
print(f"1. Accuracy (N-noqta l-3amma): {acc * 100:.2f}%")

# 3. Confusion Matrix (Fin rbeḥ o fin ghlet)
print("\n2. Confusion Matrix:")
cm = confusion_matrix(y_test, y_pred)
print(cm)

# 4. Rapport détaillé (Précision, Rappel, F1-score)
print("\n3. Classification Report:")
print(classification_report(y_test, y_pred, target_names=['Rejected (0)', 'Approved (1)']))
```

Tester le modèle sur au moins un exemple concret (données simulées ou issues du dataset) et afficher :
la prédiction (Approved / Rejected)
la probabilité associée
Sauvegarder le modèle entraîné avec pickle (model.pkl).

```python
# 1. Khassna n-3tiw l-data b-nefs t-rtib dial X
# Smiyat l-columns: Gender, Married, Dependents, Education, Self_Employed, 
# ApplicantIncome, CoapplicantIncome, LoanAmount, Loan_Amount_Term, Credit_History, Property_Area

client_1 = [[1, 1, 0, 0, 0, 5000, 2000, 150, 360, 1, 2]] # Credit_History = 1
client_2 = [[1, 0, 0, 1, 0, 2000, 0, 100, 360, 0, 0]]    # Credit_History = 0

# 2. Function bach n-testiw o n-choufou l-probabilité
def tester_client(data_client):
    prediction = model.predict(data_client)[0]
    proba = model.predict_proba(data_client)[0] # Kat-3tina [Prob_Rejected, Prob_Approved]
    
    status = "Approved ✅" if prediction == 1 else "Rejected ❌"
    confiance = proba[1] if prediction == 1 else proba[0]
    
    print(f"Résultat: {status} | Probabilité: {confiance*100:.2f}%")

# N-rj3o l-client l-jadwal sghir fih smiyat
client_1_df = pd.DataFrame(client_1, columns=X.columns)
client_2_df = pd.DataFrame(client_2, columns=X.columns)

print("Test Client 1 (Nadi):")
tester_client(client_1_df)

print("\nTest Client 2 (S3ib):")
tester_client(client_2_df)
```
